**1. Списки - базові операції**

Створіть список температур за тиждень (у градусах Цельсія) з назвою
`temperatures` і значеннями [22, 25, 19, 23, 26, 20, 21, 22].


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть перший елемент
- Виведіть останній елемент
- Додайте температуру 24
- Видаліть температуру 19
- Виведіть довжину списку
- Обчисліть суму всіх температур
- Знайдіть максимальну температуру

```python
temperatures = [22, 25, 19, 23, 26, 20, 21, 22]
```

```python
print(temperatures[0])  # Виведіть перший елемент
```

```python
print(temperatures[-1])  # Виведіть останній елемент
```

```python
temperatures.append(24)  # Додайте температуру 24
temperatures
```

```python
temperatures.remove(19)  # Видаліть температуру 19
temperatures
```

```python
print(len(temperatures))  # Виведіть довжину списку
```

```python
print(sum(temperatures))  # Обчисліть суму всіх температур
```

```python
print(max(temperatures))  # Знайдіть максимальну температуру
```


**2. Словники - базові операції**

Створіть словник про себе, який містить ключі `name`, `age`, `city` і значення, які відповідають вашему імені, віку і місту (можна вставити будь-які значення, не обовʼязково свої).


Виконайте (кожен пункт оцінюється в 0.5 бала):
- Виведіть значення ключа "name"
- Змініть значення ключа "age" на 21
- Додайте новий ключ "hobby" зі значенням "programming"
- Виведіть всі ключі
- Виведіть всі значення


```python
person = {
    "name": "Олександр",
    "age": 39,
    "city": "Дніпро"
}
```

```python
print(person["name"])   # Виведіть значення ключа "name"
```

```python
person["age"] = 21   # Змініть значення ключа "age" на 21
person
```

```python
person["hobby"] = "programming"   # Додайте новий ключ "hobby" зі значенням "programming"
person
```

```python
print(person.keys())   # Виведіть всі ключі
```

```python
print(person.values())   # Виведіть всі значення
```
